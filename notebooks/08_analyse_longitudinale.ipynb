# 08 - Analyse longitudinale

On suit l’évolution des performances en lecture dans le temps.
Dans le cadre DEPP, cela permet de mesurer les gains ou pertes sur plusieurs cohortes.


```python
import pandas as pd
import numpy as np

n = 500
rng = np.random.default_rng(42)
dates = pd.date_range('2022-01-01', periods=4, freq='Y')
df = pd.DataFrame({
    'annee': dates.year,
    'score_lecture': [56, 58, 61, 64]
})

print('Évolution du score moyen :')
print(df)
print('
Tendance moyenne :', round(df['score_lecture'].diff().mean(), 2))
```
