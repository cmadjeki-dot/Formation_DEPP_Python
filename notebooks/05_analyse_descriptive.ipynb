# 05 - Analyse descriptive

On décrit la distribution du score de lecture selon les caractéristiques du public.
Cette étape permet d’identifier les écarts de niveau et les premiers signaux explicatifs.


```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.data.loader import load_csv, data_path

df = load_csv(data_path('data', 'interim', 'etude_lecture_6e_clean.csv'))
print('Moyenne score_lecture :', round(df['score_lecture'].mean(), 2))
print('Écart-type :', round(df['score_lecture'].std(), 2))

print('
Moyenne par PCS :')
print(df.groupby('pcs')['score_lecture'].mean().sort_values(ascending=False).round(2))

print('
Moyenne par sexe :')
print(df.groupby('sexe')['score_lecture'].mean().round(2))

print('
Moyenne par retard :')
print(df.groupby('retard')['score_lecture'].mean().round(2))
```
