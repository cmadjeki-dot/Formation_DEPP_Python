# 03 - Contrôle qualité statistique

Avant toute analyse, il faut vérifier :
- présence de doublons ;
- valeurs manquantes ;
- cohérence des types ;
- valeurs aberrantes.


```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.data.loader import load_csv, data_path
from src.quality.check import duplicate_count, missing_values_report, summarize_quality

df = load_csv(data_path('data', 'raw', 'etude_lecture_6e.csv'))
print('Doublons :', duplicate_count(df))
print('
Valeurs manquantes par variable :')
print(missing_values_report(df))
print('
Synthèse qualité :')
print(summarize_quality(df))

score = df['score_lecture']
print('
Distribution du score lecture :')
print(score.describe())
print('Valeurs aberrantes (score < 0 ou > 100) :', int(((score < 0) | (score > 100)).sum()))
```
