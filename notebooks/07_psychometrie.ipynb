# 07 - Psychométrie

On construit une batterie d’items de lecture et on vérifie leur cohérence.
Le point clé est le calcul de l’alpha de Cronbach.


```python
import pandas as pd
import numpy as np

rng = np.random.default_rng(42)
n = 1500
latent = rng.normal(0, 1, n)
items = pd.DataFrame({
    'item_1': np.clip(latent + rng.normal(0, 0.8, n), 0, 100),
    'item_2': np.clip(latent + rng.normal(0, 0.8, n), 0, 100),
    'item_3': np.clip(latent + rng.normal(0, 0.9, n), 0, 100),
    'item_4': np.clip(latent + rng.normal(0, 0.8, n), 0, 100),
})

correlations = items.corr()
print('Matrice de corrélations :')
print(correlations.round(3))

n_items = items.shape[1]
item_var = items.var(axis=0, ddof=1)
total_var = items.sum(axis=1).var(ddof=1)
alpha = (n_items / (n_items - 1)) * (1 - item_var.sum() / total_var)
print('
Alpha de Cronbach :', round(alpha, 3))
```
