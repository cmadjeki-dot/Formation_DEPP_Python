# 04 - Traitement des valeurs manquantes

Le but est d’éliminer les biais liés à la non-réponse sans supprimer trop d’informations.
On utilise ici une imputation simple et transparente, avec la médiane pour les variables continues.


```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.data.loader import load_csv, save_csv, data_path

df = load_csv(data_path('data', 'raw', 'etude_lecture_6e.csv'))

numeric_cols = ['score_lecture', 'ressources_num', 'age']
for col in numeric_cols:
    if col in df.columns:
        median_value = df[col].median()
        df[col] = df[col].fillna(median_value)

categorical_cols = ['sexe', 'pcs', 'academie', 'type_etab']
for col in categorical_cols:
    if col in df.columns:
        mode_value = df[col].mode(dropna=True)
        if not mode_value.empty:
            df[col] = df[col].fillna(mode_value.iloc[0])

print('Valeurs manquantes après traitement :')
print(df.isna().sum())

clean_path = data_path('data', 'interim', 'etude_lecture_6e_clean.csv')
save_csv(df, clean_path)
print('Base nettoyée exportée vers :', clean_path)
```
