# 13 - Oral final DEPP

Ce notebook sert de support de présentation orale.
Il reprend les messages principaux à délivrer devant un jury ou un décideur.


## 1. Problématique

Nous étudions l’impact du contexte socio-économique sur les performances en lecture des élèves de 6e.
L’enjeu est de mesurer les écarts de maîtrise et d’identifier les leviers de politique éducative.


## 2. Méthodologie

La démarche suit la logique DEPP :
- définition du besoin ;
- protocole statistique ;
- construction de la base ;
- contrôle qualité ;
- imputation ;
- analyse descriptive ;
- régression et psychométrie ;
- restitution décisionnelle.


## 3. Analyse quantitative

Les résultats montrent que :
- les élèves issus des milieux favorisés obtiennent de meilleurs scores ;
- le retard scolaire est associé à une baisse significative des performances ;
- le type d’établissement et les ressources numériques jouent un rôle explicatif ;
- les écarts restent présents, même après prise en compte des caractéristiques observées.


## 4. Interprétation

Le contexte socio-économique demeure un facteur majeur de différenciation des performances.
Les dispositifs existants réduisent partiellement les écarts, mais ne les éliminent pas.
Cela confirme l’importance d’une politique éducative ciblée et précoce.


## 5. Recommandations

- renforcer le soutien aux élèves en retard ;
- cibler les zones et établissements les plus fragiles ;
- améliorer l’accès aux ressources numériques ;
- poursuivre le suivi longitudinal des performances.


## 6. Phrase de conclusion

Les analyses montrent que le contexte socio-économique reste un déterminant majeur des performances en lecture. Malgré les dispositifs existants, des écarts significatifs persistent entre groupes sociaux et territoires, ce qui appelle une action publique ciblée et précoce.


## 7. Script oral

"Nous avons mené une étude sur l’impact du contexte socio-économique sur les performances en lecture des élèves de 6e. À partir d’une base simulée cohérente avec les exigences d’une production statistique publique, nous avons suivi toute la chaîne de travail : définition du besoin, protocole, collecte, contrôle qualité, correction, analyse descriptive, modélisation et restitution. Les résultats montrent que les élèves issus des milieux favorisés obtiennent des scores supérieurs, que le retard scolaire est associé à des performances plus faibles et que les écarts persistent même après prise en compte des caractéristiques observées. L’enjeu principal est donc de mieux comprendre la transmission des inégalités scolaires et d’orienter les politiques publiques vers une aide ciblée et précocement."
