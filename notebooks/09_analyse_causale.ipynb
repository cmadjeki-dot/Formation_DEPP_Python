# 09 - Analyse causale

On simule un dispositif d’aide et on mesure son effet sur les performances.
L’idée est d’étudier le lien entre traitement et résultats, tout en contrôlant le contexte social.


```python
import pandas as pd

df = pd.DataFrame({
    'traitement': ['A', 'A', 'B', 'B'],
    'score': [52, 54, 58, 60],
    'pcs': ['faible', 'faible', 'favorise', 'favorise']
})

summary = df.groupby('traitement')['score'].mean()
print('Effet brut du dispositif :')
print(summary)

print('
Effet moyen :', round(summary.B - summary.A, 2))
```
